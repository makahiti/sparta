### 문제 4


T(n) = T(n/2) + 1,\quad T(1) = 1
\begin{aligned}
&= T(n/2) + 1 \\
&= T(n/4) + 1 \\
&= T(n/8) + 1 \\
&= T(n/2^k) + 1 \quad \rightarrow \quad 2^k = n
\end{aligned}
O(\log N)


### 문제 12

**명제:**  
\( n^2 \)이 3의 배수이면, \( n \)은 3의 배수임을 증명하라.

---

**대우:**  
\( n \)이 3의 배수가 아니라면, \( n^2 \)도 3의 배수가 아니다.

---

\( n = 3k + 1 \) 이라면,

\[
n^2 = (3k + 1)^2 = 9k^2 + 6k + 1 = 3(3k^2 + 2k) + 1
\]

→ 3의 배수가 아니다.

---

\( n = 3k + 2 \) 이라면,

\[
n^2 = (3k + 2)^2 = 9k^2 + 12k + 4 = 3(3k^2 + 4k + 1) + 1
\]

→ 3의 배수가 아니다.

---

**결론:**  
대우가 참 → 원래 명제도 참
