아래 그림과 같은 미로가 있다. 16*16 행렬의 형태로 만들어진 미로에서 흰색 바탕은 길, 노란색 바탕은 벽을 나타낸다.

가장 좌상단에 있는 칸을 (0, 0)의 기준으로 하여, 가로방향을 x 방향, 세로방향을 y 방향이라고 할 때, 미로의 시작점은 (1, 1)이고 도착점은 (13, 13)이다.

주어진 미로의 출발점으로부터 도착지점까지 갈 수 있는 길이 있는지 판단하는 프로그램을 작성하라.

아래의 예시에서는 도달 가능하다.

```python
T = int(input())
for tc in range(1,1+T):
    N = int(input())
    matrix = [list(map(int, input())) for _ in range(N)]
 
    si, sj = 0, 0
    for i in range(N):
        for j in range(N):
            if matrix[i][j] == 2:
                si, sj = i, j
 
    di = [0,0,1,-1]
    dj = [1,-1,0,0]
 
    def dfs(si,sj):
        visited = [[0]*N for _ in range(N)]
        visited[si][sj] = 1
        stack = []
        i, j = si, sj
 
        while True:
            if matrix[i][j] == 3:
                return 1
             
            for d in range(4):
                ni = i + di[d]
                nj = j + dj[d]
                if 0 <= ni < N and 0 <= nj < N and not visited[ni][nj] and matrix[ni][nj] != 1:
                    visited[ni][nj] = 1
                    stack.append((i,j))
                    i, j = ni, nj
                    break
 
            else:
                if stack:
                    i, j = stack.pop()
                else:
                    break
         
        return 0
     
    print(f'#{tc} {dfs(si,sj)}')
```

```python
T = int(input())

for tc in range(1, T + 1):
    # 미로의 크기 N
    N = int(input())

    # 미로의 정보
    # 0 : 이동가능
    # 1 : 벽(이동불가)
    # 2 : 시작지점
    # 3 : 출구지점
    maze = [list(map(int, input())) for _ in range(N)]

    # 시작지점 찾기
    si, sj = 0, 0

    # 행 우선순회로 일단 2차원 배열 순회해서 시작지점 찾기
    for i in range(N):
        for j in range(N):
            if maze[i][j] == 2:
                si, sj = i, j

    di = [-1, 1, 0, 0]
    dj = [0, 0, -1, 1]


    def dfs(si, sj):

        visited = [[0] * N for _ in range(N)]
        # visited[5][5] = 1 : 5행5열 방문 한적 있다.
        # visited[1][2] = 0 : 1행2열 방문 한적 없다.

        stack = []

        # 시작지점 방문 체크
        visited[si][sj] = 1

        # 현재 방문중인 위치 i, j
        i, j = si, sj

        while True:

            # 현재 있는 위치가 출구면 반복중단
            if maze[i][j] == 3:
                # 출구 찾음, 정답 1
                return 1

            # 현재 위치에서 인접한 4방향 탐색
            for d in range(4):
                ni = i + di[d]
                nj = j + dj[d]

                # (i,j) 위치와 인접한 (ni,nj)위치가 이동 가능한가???
                # 2차원 배열 인덱스 범위 안
                # 방문한적 없어야 함
                # 벽이 아니어야 함
                if 0 <= ni < N and 0 <= nj < N and not visited[ni][nj] and maze[ni][nj] != 1:
                    # ni,nj 방문체크
                    visited[ni][nj] = 1
                    # 돌아올 위치 기억
                    stack.append((i, j))
                    # ni,nj 이동
                    i, j = ni, nj
                    break
            else:
                # 중간에 break 한적이 없다 => 갈곳을 발견하지 못했다. 이전 방문지점으로 돌아감
                if stack:
                    i, j = stack.pop()
                else:
                    # 스택 비었으면 갈수 있는곳 다 가본거
                    break

        # 반복문 다 돌았는데 return 1 못하면 출구를 찾지 못함
        return 0


    print(f"#{tc} {dfs(si, sj)}")

"""

3
5
13101
10101
10101
10101
10021
5
10031
10111
10101
10101
12001
5
00013
01110
21000
01111
00000

"""
```

```python
# 테스트 케이스 개수 입력
T = int(input())

# 각 테스트 케이스 처리
for tc in range(1, T + 1):

    # 미로의 크기 (N x N)
    N = int(input())

    # -----------------------------
    # 미로 입력
    # -----------------------------
    # input()으로 한 줄을 문자열로 받는다
    # 예: "13101"
    # map(int, input()) → 문자 하나씩 정수로 변환
    # 결과: [1, 3, 1, 0, 1]
    maze = [list(map(int, input())) for _ in range(N)]

    # -----------------------------
    # 1️⃣ 출발 지점 찾기 (값이 2인 칸)
    # -----------------------------
    # 출발 좌표를 저장할 변수
    sy, sx = 0, 0

    # 미로 전체를 순회하며 출발점 탐색
    for i in range(N):
        for j in range(N):
            if maze[i][j] == 2:
                sy, sx = i, j   # 출발 위치 저장

    # -----------------------------
    # 2️⃣ 방문 여부 확인용 배열
    # -----------------------------
    # visited[y][x] == 1 → 이미 방문한 칸
    # visited[y][x] == 0 → 아직 방문 안 한 칸
    visited = [[0] * N for _ in range(N)]

    # -----------------------------
    # 3️⃣ 상, 하, 좌, 우 이동용 방향 배열
    # -----------------------------
    # y좌표 변화량
    di = [-1, 1, 0, 0]
    # x좌표 변화량
    dj = [0, 0, -1, 1]

    # -----------------------------
    # 4️⃣ BFS용 큐 (리스트로 구현)
    # -----------------------------
    # 큐에는 (y좌표, x좌표, 지금까지 지나온 칸 수)를 저장
    queue = []

    # 시작 지점 삽입
    # 시작점도 '지나간 칸'에 포함되므로 거리 = 1
    queue.append((sy, sx, 1))
    visited[sy][sx] = 1   # 시작 지점 방문 표시

    # 도착 지점에 도달했을 때의 최소 칸 수
    # 도착하지 못하면 0이 유지됨
    answer = 0

    # -----------------------------
    # 5️⃣ BFS 탐색 시작
    # -----------------------------
    while queue:

        # 큐의 맨 앞 요소 꺼내기
        # 리스트이지만 pop(0)을 사용해 FIFO 구조를 만든다
        y, x, dist = queue.pop(0)

        # -------------------------
        # 현재 위치가 도착 지점(3)인 경우
        # -------------------------
        # BFS는 가까운 거리부터 탐색하므로
        # 처음 도착했을 때의 dist가 최소 거리
        if maze[y][x] == 3:
            answer = dist
            break   # 더 탐색할 필요 없음

        # -------------------------
        # 현재 위치에서 상, 하, 좌, 우 이동
        # -------------------------
        for d in range(4):

            # 다음 위치 계산
            ny = y + di[d]
            nx = x + dj[d]

            # ---------------------
            # 이동 가능 조건 검사
            # ---------------------
            # 1) 미로 범위 안인지
            # 2) 아직 방문하지 않았는지
            # 3) 벽(1)이 아닌지
            if 0 <= ny < N and 0 <= nx < N:
                if not visited[ny][nx] and maze[ny][nx] != 1:

                    # 해당 칸을 방문 처리
                    # 큐에 넣기 전에 방문 표시를 해야
                    # 중복 방문을 막을 수 있음
                    visited[ny][nx] = 1

                    # 다음 위치를 큐에 추가
                    # 한 칸 이동했으므로 거리 +1
                    queue.append((ny, nx, dist + 1))

    # -----------------------------
    # 6️⃣ 결과 출력
    # -----------------------------
    print(f"#{tc} {answer}")
```
